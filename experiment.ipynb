# Analyse data from experiments

## Loading

```python
import pandas as pd
dataset = 'mnist'
df = pd.read_csv(f'outputs/experiments-{dataset}.csv')
```

```python
df.head(20)
```

## Plot results

```python
import matplotlib.pyplot as plt

x = df['DG sparsity']
y = df['CA3 sparsity']
cols = ['Average correlation', 'Validation error']

for i in range(len(cols)):
    plt.scatter(x, y, c=df[cols[i]], cmap='viridis')
    plt.xlabel('DG sparsity')
    plt.ylabel('CA3 sparsity')
    plt.colorbar(label=cols[i])
    plt.show()
```
